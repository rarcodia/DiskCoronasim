```python
import DiskCoronaSim
import Plot
from Plot import load_data_specific, plot_data
```

# 1) Example with a single object

```python
mass=1e8
mdot=0.3
photon_index=1.9
spin=0
DiskCoronaSim.runmodel(mass,mdot,photon_index=photon_index, spin_input=spin)
json_example=load_data_specific(mass,mdot,photon_index=photon_index, spin_input=spin)
plot_data(json_example[0], json_example[1])
```

# 2) Example with multiple objects

- Note that for visualization purposes it is easier to plot a single mass with a range of accretion rates

```python
json_files=[]
json_files_names=[]
for i in 1e8, :
    for j in 0.1, 0.3, 0.7, 1.0 :
        DiskCoronaSim.runmodel(i, j)
        jsons, names = load_data_specific(i, j)
        json_files.append(jsons)
        json_files_names.append(names)
plot_data(json_files, json_files_names)
```
