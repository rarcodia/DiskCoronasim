# 2) Example with multiple objects

- Note that for visualization purposes it is easier to plot a single mass with a range of accretion rates

```python
json_files=[]
json_files_names=[]
for i in 1e8, :
    for j in 0.1, 0.3, 0.5, 0.7 :
        DiskCoronaSim.runmodel(i, j)
        jsons, names = load_data_specific(i, j)
        json_files.append(jsons)
        json_files_names.append(names)
plot_data(json_files, json_files_names)
```
